```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path("../data/raw/Telco-Customer-Churn.csv")
df = pd.read_csv(DATA_PATH)
print("Dataset_shape",df.shape)
display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)


print("\nChurn distribution:")
print(df["Churn"].value_counts(dropna=False))

print("\nMissing values:")
print(df.isnull().sum())
```

```python

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print(df.dtypes)
```

```python
for col in df.select_dtypes(include="object").columns:
    print(f"\n{col}:")
    print(df[col].value_counts().head(10))
```

```python
print("Empty strings")
print((df.astype("string").apply(lambda col:col.str.strip().eq("")).sum()))

print("\nWhite space-only values")
print((df.astype("string").apply(lambda col:col.str.fullmatch(r"\s+").fillna(False)).sum()))
```

```python
print("Duplicate rows:", df.duplicated().sum())

print("Duplicate customer IDs:", df["customerID"].duplicated().sum())

print("\nUnique churn values:")
print(df["Churn"].unique())
```

```python
print(df["TotalCharges"].dtype)
print(df["TotalCharges"].head(10).tolist())
```

```python
empty_charges = df["TotalCharges"].astype(str).str.strip().eq("")
print("Empty total charges rows : ",empty_charges.sum())
display(
  df.loc[
    empty_charges,["customerID","tenure","MonthlyCharges","Churn"]
  ]
)
```

```python
import pandas as pd

charges_numeric = pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Values that cannot be converted:", charges_numeric.isna().sum())
print("Tenure distribution for missing charges:")
print(df.loc[charges_numeric.isna(), "tenure"].value_counts())
```
